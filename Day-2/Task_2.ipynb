```python
def get_cake_price(cake: str) -> int:
    prices = {
        "chocolate": 500,
        "vanilla": 400,
        "plum": 600
    }

    return prices.get(cake.lower(), 0)


price = get_cake_price("Chocolate")

print("Cake price:", price)
```

```python
def my_decorator(func):
    def wrapper():
        print("Function started")
        func()
        print("Function finished")
    return wrapper


@my_decorator
def greet():
    print("Hello buddy!")


greet()
```

```python
def get_cake_price(cake: str) -> int:
    prices = {
        "chocolate": 500,
        "vanilla": 400,
        "plum": 600
    }

    try:
        return prices[cake.lower()]

    except KeyError:
        print("Sorry, this cake is not available.")
        return 0


print(get_cake_price("chocolate"))
print(get_cake_price("strawberry"))
```
